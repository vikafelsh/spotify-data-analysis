```python
import matplotlib.pyplot as plt
```

```python
plt.style.use('dark_background')

top10 = data.nlargest(10, 'total_streams_billions')

plt.figure(figsize=(12, 6))
plt.barh(top10['song_title'], top10['total_streams_billions'], color='purple')

plt.title('Top 10 Most Streamed Spotify Songs')
plt.xlabel('Streams (billions)')
plt.ylabel('Song Title')
plt.gca().invert_yaxis()
plt.tight_layout()
plt.show()
```

```python
import seaborn as sns
```

```python
sns.set_theme(style='darkgrid')

plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=data,
    x='bpm',
    y='energy',
    hue='explicit',
    palette=['royalblue', 'tomato'],
    s=90
)

plt.title('BPM vs Energy')
plt.xlabel('Beats Per Minute (BPM)')
plt.ylabel('Energy')
plt.legend(title='Explicit')
plt.show()
```

```python
import plotly.express as px
```

```python
plt.style.use('ggplot')

genres = data['primary_genre'].value_counts().head(5)

plt.figure(figsize=(8, 8))
plt.pie(
    genres,
    labels=genres.index,
    autopct='%1.1f%%',
    startangle=90
)

plt.title('Top 5 Music Genres')
plt.show()
```

```python
plt.style.use('bmh')

streams = data.groupby('release_year')['total_streams_billions'].mean()

plt.figure(figsize=(11, 5))
plt.plot(
    streams.index,
    streams.values,
    marker='o',
    color='royalblue',
    linewidth=2
)

plt.title('Average Spotify Streams by Release Year')
plt.xlabel('Release Year')
plt.ylabel('Average Streams (billions)')
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
fig = px.bar(
    data,
    x='song_title',
    y='total_streams_billions',
    color='primary_genre',
    animation_frame='release_year',
    title='Spotify Songs by Release Year'
)
fig.show()
```

```python
fig = px.scatter(
    data,
    x='danceability',
    y='energy',
    color='primary_genre',
    animation_frame='release_year',
    hover_name='song_title',
    range_x=[0, 1],
    range_y=[0, 1],
    title='Danceability vs Energy',
    template='plotly_dark'
)

fig.show()
```
